```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [4, 5, 6])
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Sales"], bins=10, edgecolor="black")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["Sales"])
plt.title("Box Plot of Sales")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df["Category"].value_counts().plot(kind="bar", color="skyblue")
plt.title("Products by Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

sales_trend = df.groupby("Order Date")["Sales"].sum()

plt.figure(figsize=(10,5))
plt.plot(sales_trend.index, sales_trend.values, marker="o")
plt.title("Sales Trend Over Time")
plt.xlabel("Order Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

Technology products generated the highest sales.

Furniture had moderate sales compared to other categories.

The box plot shows a few high-value sales (outliers).

Sales vary over time based on order dates.

Some numerical columns have positive correlations.
