```python
query = """
SELECT "Order ID", "Customer Name", Category, Sales, Profit
FROM superstore
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT "Order ID", "Customer Name", Category, Sales, Profit
FROM superstore
WHERE Sales > 1000
ORDER BY Sales DESC;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT "Order ID", "Customer Name", Sales, Profit
FROM superstore
ORDER BY Sales DESC
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT Category,
       SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Category
ORDER BY Total_Profit DESC;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY Category
HAVING SUM(Sales) > 100000
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT "Sub-Category",
       SUM(Sales) AS Total_Sales
FROM superstore
GROUP BY "Sub-Category"
ORDER BY Total_Sales DESC
LIMIT 5;
"""

top_5_subcategories = pd.read_sql(query, conn)
top_5_subcategories
```

```python
query = """
SELECT Region,
       SUM(Sales) AS Total_Sales,
       SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Region
ORDER BY Total_Sales DESC;
"""

sales_by_region = pd.read_sql(query, conn)
sales_by_region
```

```python
query = """
SELECT "Customer Name",
       SUM(Sales) AS Total_Sales,
       SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY "Customer Name"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

top_customers = pd.read_sql(query, conn)
top_customers
```

```python
query = """
CREATE TABLE IF NOT EXISTS region_info (
    Region TEXT PRIMARY KEY,
    Region_Type TEXT
);
"""

conn.execute(query)

region_data = [
    ("Central", "Domestic"),
    ("East", "Domestic"),
    ("South", "Domestic"),
    ("West", "Domestic")
]

conn.executemany(
    "INSERT OR IGNORE INTO region_info (Region, Region_Type) VALUES (?, ?)",
    region_data
)

conn.commit()

print("region_info table created successfully!")
```

```python
pd.read_sql("SELECT * FROM region_info;", conn)
```

```python
query = """
SELECT
    s.Region,
    r.Region_Type,
    SUM(s.Sales) AS Total_Sales,
    SUM(s.Profit) AS Total_Profit
FROM superstore s
INNER JOIN region_info r
    ON s.Region = r.Region
GROUP BY s.Region, r.Region_Type
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT
    "Order ID",
    "Customer Name",
    Sales,
    Profit
FROM superstore
WHERE Sales > (
    SELECT AVG(Sales)
    FROM superstore
)
ORDER BY Sales DESC;
"""

above_average_sales = pd.read_sql(query, conn)
above_average_sales.head(10)
```

```python
query = """
WITH subcategory_sales AS (
    SELECT
        "Sub-Category",
        SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY "Sub-Category"
)
SELECT
    "Sub-Category",
    Total_Sales
FROM subcategory_sales
ORDER BY Total_Sales DESC
LIMIT 5;
"""

top_subcategories = pd.read_sql(query, conn)
top_subcategories
```

```python
query = """
SELECT
    "Order ID",
    "Customer Name",
    Sales,
    ROW_NUMBER() OVER (ORDER BY Sales DESC) AS Sales_Row_Number
FROM superstore
ORDER BY Sales DESC
LIMIT 10;
"""

result = pd.read_sql(query, conn)
result
```

```python
query = """
WITH customer_sales AS (
    SELECT
        "Customer Name",
        SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY "Customer Name"
)
SELECT
    "Customer Name",
    Total_Sales,
    RANK() OVER (ORDER BY Total_Sales DESC) AS Sales_Rank
FROM customer_sales
ORDER BY Sales_Rank
LIMIT 10;
"""

customer_ranking = pd.read_sql(query, conn)
customer_ranking
```

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS Month,
        SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY Month
)
SELECT
    Month,
    Total_Sales,
    LEAD(Total_Sales) OVER (ORDER BY Month) AS Next_Month_Sales
FROM monthly_sales
ORDER BY Month;
"""

next_month = pd.read_sql(query, conn)
next_month
```

```python
query = """
CREATE VIEW IF NOT EXISTS regional_sales AS
SELECT
    Region,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    SUM(Quantity) AS Total_Quantity
FROM superstore
GROUP BY Region;
"""

conn.execute(query)
conn.commit()

print("View created successfully!")
```
